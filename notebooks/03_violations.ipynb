# 03 Parking violations (2023)
Can fines be tied to parking areas, or only to streets?

```python
from pathlib import Path
import geopandas as gpd, pandas as pd, matplotlib.pyplot as plt

RAW = sorted(Path("../data/raw").iterdir())[-1]  # latest snapshot
print("snapshot:", RAW.name)
areas = gpd.read_file(RAW / "parking_areas_3879.geojson")
fines = gpd.read_file(RAW / "violations_2023_3879.geojson")
```

```python
fines["virhemaksun_vaihe"].value_counts()
```

## Location precision
Many fines share one address point.

```python
fines.geometry.to_wkt().value_counts().head(10)
```

```python
fines = fines[fines["todellinen_sijainti"] == "Kyllä"]
near = gpd.sjoin_nearest(fines, areas[["geometry"]], distance_col="dist")
near["dist"].describe(percentiles=[.1, .25, .5, .75, .9])
```

## Reasons and streets

```python
fines["virheen_paasyy_ja_paaluokka"].value_counts().head(15)
```

```python
fines["street"] = fines["osoite"].str.replace(r"\s*\d.*$", "", regex=True)
fines["street"].value_counts().head(20)
```
